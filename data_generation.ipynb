```python
import numpy as np
import pandas as pd

np.random.seed(42)
print("Libraries loaded ✅")
```

```python
n_days = 365
start_date = pd.Timestamp('2024-01-01')
categories = ['Food', 'Transport', 'Shopping', 'Utilities', 'Entertainment', 'Healthcare']
cat_params = {
    'Food':          {'mean': 400,  'std': 80},
    'Transport':     {'mean': 200,  'std': 50},
    'Shopping':      {'mean': 600,  'std': 200},
    'Utilities':     {'mean': 1200, 'std': 100},
    'Entertainment': {'mean': 300,  'std': 100},
    'Healthcare':    {'mean': 150,  'std': 60},
}

print("Config set ✅")
```

```python
rows = []

for day in pd.date_range(start_date, periods=n_days):
    n_txns = np.random.randint(2, 5)      # 2-4 transactions per day
    for _ in range(n_txns):
        cat = np.random.choice(categories)
        p = cat_params[cat]
        amount = max(10, np.random.normal(p['mean'], p['std']))
        rows.append({
            'date':       day,
            'category':   cat,
            'amount':     round(amount, 2),
            'is_anomaly': 0                # 0 = normal
        })

df = pd.DataFrame(rows)
print(f"Generated {len(df)} transactions ✅")
df.head(10)
```

```python
anomaly_indices = np.random.choice(df.index, size=20, replace=False)
df.loc[anomaly_indices, 'amount'] *= np.random.uniform(4, 8, size=20)
df.loc[anomaly_indices, 'amount']  = df.loc[anomaly_indices, 'amount'].round(2)
df.loc[anomaly_indices, 'is_anomaly'] = 1
print(f"Anomalies injected: {df['is_anomaly'].sum()} ✅")
print(df[df['is_anomaly'] == 1][['date', 'category', 'amount']].head(10))
```

```python
import os
os.makedirs('../data', exist_ok=True)
df.to_csv('../data/transactions.csv', index=False)
print("Saved to data/transactions.csv ✅")
print(f"\nDataset shape: {df.shape}")
print(f"\nCategory breakdown:\n{df['category'].value_counts()}")
```

```python
print("=== DATASET SUMMARY ===")
print(f"Total transactions  : {len(df)}")
print(f"Date range          : {df['date'].min().date()} → {df['date'].max().date()}")
print(f"Normal transactions : {(df['is_anomaly']==0).sum()}")
print(f"Anomalies           : {(df['is_anomaly']==1).sum()}")
print(f"\nAmount stats:")
print(df['amount'].describe().round(2))
```
