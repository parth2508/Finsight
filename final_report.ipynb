```python
import pandas as pd
import numpy as np

df = pd.read_csv('../data/transactions_flagged.csv', parse_dates=['date'])
print("Data loaded ✅")
```

```python
flagged = df[df['flag_iqr'] == 1]
top     = flagged.loc[flagged['amount'].idxmax()]

print("=" * 55)
print("   PERSONAL FINANCE ANOMALY REPORT — 2024")
print("=" * 55)

print(f"\n📊 OVERVIEW")
print(f"   Total Transactions Analyzed : {len(df):,}")
print(f"   Date Range                  : {df['date'].min().date()} → {df['date'].max().date()}")
print(f"   Total Annual Spend          : ₹{df['amount'].sum():,.2f}")
print(f"   Average Daily Spend         : ₹{df.groupby('date')['amount'].sum().mean():,.2f}")

print(f"\n🚨 ANOMALY SUMMARY")
print(f"   Anomalies Detected (IQR)    : {len(flagged)}")
print(f"   Anomaly Rate                : {len(flagged)/len(df)*100:.2f}%")
print(f"   Most Affected Category      : {flagged['category'].value_counts().idxmax()}")

print(f"\n💥 HIGHEST ANOMALOUS TRANSACTION")
print(f"   Amount    : ₹{top['amount']:,.2f}")
print(f"   Date      : {top['date'].date()}")
print(f"   Category  : {top['category']}")

print(f"\n📅 MONTHLY ANOMALY BREAKDOWN")
monthly = df.groupby(df['date'].dt.to_period('M'))['flag_iqr'].sum()
for month, count in monthly.items():
    bar = '█' * int(count)
    print(f"   {month} : {bar} ({count})")

print(f"\n📂 CATEGORY RISK RANKING")
cat_risk = df.groupby('category')['flag_iqr'].sum().sort_values(ascending=False)
for cat, count in cat_risk.items():
    print(f"   {cat:<15} : {count} anomalies")

print("\n" + "=" * 55)
print("   END OF REPORT")
print("=" * 55)
```

```python
import sys
from io import StringIO

# Capture print output
old_stdout = sys.stdout
sys.stdout = buffer = StringIO()

# Run report again to capture it
flagged = df[df['flag_iqr'] == 1]
top     = flagged.loc[flagged['amount'].idxmax()]

print("=" * 55)
print("   PERSONAL FINANCE ANOMALY REPORT — 2024")
print("=" * 55)
print(f"\nTotal Transactions : {len(df):,}")
print(f"Total Spend        : ₹{df['amount'].sum():,.2f}")
print(f"Anomalies Detected : {len(flagged)}")
print(f"Anomaly Rate       : {len(flagged)/len(df)*100:.2f}%")
print(f"Most Affected      : {flagged['category'].value_counts().idxmax()}")
print(f"Highest Anomaly    : ₹{top['amount']:,.2f} on {top['date'].date()}")
print("=" * 55)

sys.stdout = old_stdout
report_text = buffer.getvalue()

with open('../data/anomaly_report.txt', 'w', encoding='utf-8') as f:
    f.write(report_text)

print("Report saved to data/anomaly_report.txt ✅")
print(report_text)
```
